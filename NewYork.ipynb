```python
def is_leap_year(year):
    if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
        return True
    return False

# Test
print(is_leap_year(2015))  
print(is_leap_year(2016))  

```

```python
def count_even_odd(numbers):
    even = 0
    odd = 0

    for num in numbers:
        if num % 2 == 0:
            even += 1
        else:
            odd += 1

    return even, odd

print(count_even_odd([1,2,3,4,5,6,8,9,10]))
```

```python
def find_second_largest(numbers):
    unique_numbers = list(set(numbers))
    unique_numbers.sort(reverse=True)
    
    if len(unique_numbers) < 2:
        return None
    
    return unique_numbers[1]


print(find_second_largest([10, 20, 30, 40, 40]))
```

```python
df.columns = df.columns.str.replace(" ", "_")
print(df.columns)
```

```python
df["SALE_PRICE"] = pd.to_numeric(
    df["SALE_PRICE"].replace("-", pd.NA)
)
median_price = df["SALE_PRICE"].median()

df["SALE_PRICE"].fillna(median_price, inplace=True)
```

```python
avg_units = df["TOTAL_UNITS"].mean()
print(avg_units)
```

```python
chart = df.groupby("YEAR_BUILT")["SALE_PRICE"].mean()

plt.figure(figsize=(12,5))
plt.plot(chart.index, chart.values)

plt.title("Year Built vs Sale Price")
plt.xlabel("Year Built")
plt.ylabel("Average Sale Price")

plt.grid(True)
plt.show()
```
